```python
schema = StructType([
    StructField("TransactionID", StringType(), True),
    StructField("CustomerID", StringType(), True),
    StructField("CustomerDOB", StringType(), True),
    StructField("CustGender", StringType(), True),
    StructField("CustLocation", StringType(), True),
    StructField("CustAccountBalance", StringType(), True),
    StructField("TransactionDate", StringType(), True),
    StructField("TransactionTime", StringType(), True),
    StructField("TransactionAmountINR", StringType(), True)
])
```

```python
raw_bt = spark.read.csv("/Volumes/workspace/default/landing_zone/bank_transactions.csv", header=True, schema=schema)
```

```python
raw_bt.printSchema()
```

```python
bronze_bt = raw_bt.withColumn("ingestion_timestamp", current_timestamp()).withColumn("source_file", col("_metadata.file_path"))
```

```python
bronze_bt.printSchema()
```

```python
bronze_bt.show(5)
```
